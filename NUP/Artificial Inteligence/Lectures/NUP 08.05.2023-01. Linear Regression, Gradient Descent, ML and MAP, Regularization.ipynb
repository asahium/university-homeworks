## First step: ChatGPT
1. We open a new dialogue with [ChatGPT](https://chat.openai.com/) and ask to generate some names of the most popular characters, for example, from the movie or series. The more popular the project was, the better the result will be with further processing by neural networks.
2. Having received a list of names, we ask ChatGPT how they can look and what they can wear, for example, at a fashion show of a particular brand.

Here is an example of a text query in English:
> Now, act as a N (substitute Balenciaga or any other brand) fashion designer with 20 years of experience. I want you to dress each of these N (number of characters) characters in a unique N (brand name) piece for a runway show. Just write the name of the character and the piece. Remember, itʼs a N (year) fashion show.

3. We teach ChatGPT to understand prompts for Midjourney - text descriptions of pictures that we will generate in the future. To do this, write the following message to the chatbot about one of the characters:
> N (year) screengrab of N (male, female) model N (character name) wearing a grotesque N (clothing color and brand name) dress, fashion movie scene, N (brand name) commercial —ar 3:2 —v 5. Just say "Yes" if you understand.

In order not to make similar text queries for all characters, we ask for help from ChatGPT:

> I want you to rewrite the prompt but for every character, with the clothes you chose for them. Remember to write the prompt exactly as I pasted it, with the same format and parameters. Only change the gender of the model, the name of the character, and the description of the clothing. Do you understand? Please ask if you need more information.

## Second step: Midjourney
1. Having received a list of text requests (prompts) from ChatGPT, go to [Midjourney discord server](https://discord.com/invite/midjourney) to generate images with characters based on them.
2. In one of the chats on the Midjourney discord server, we send prompts one by one. When the neural network generates four images, choose one or more of the ones you like the most and download them to your device. Repeat with each prompt.

## Third step: YouTube and converter
1. To generate a voice that is as close as possible to a real one, a neural network needs a source from which to learn. For this, a YouTube video is suitable, in which the actor playing the character (for example, Daniel Radcliffe playing Harry Potter) is talking. Of course, it is important that this is not a dialogue, but a monologue.
2. Cut the audio using any of the converter services that suit you, which will convert YouTube videos to mp3 files.

Repeat this for all characters.

## Fourth step: ElevenLabs
1. Go to the [ElevenLabs](https://beta.elevenlabs.io/) service (note that it is still in beta version), with the help of which we will generate voices for the characters. You need to add a pre-cut audio track with the hero's voice and come up with a text that he will pronounce.

Most often, the authors of such videos used the signature phrases of the characters. Start audio generation, wait for the result and download it.

Repeat the operation with all the characters that will speak in your video.

## Fifth step: D-ID and CapCut
By the final stage, we have generated photos of the characters and the voices of their neural network "clones". Now we open the [D-ID service](https://www.d-id.com/), with which we will create an animated clip.

1. D-ID offers instructions for beginners, but you can do without it. Load each image generated by Midjourney in turn.
2. Choose whether the character will be silent or speak. If you want your character to say the phrase you created with ElevenLabs, upload an audio file from the menu that opens on the right. You can also choose the duration of the animation in the video.
3. Do the same with all the characters, and then glue the resulting videos. To do this, you can use the free [CapCut](https://www.capcut.com/) service.
4. To create flashes, you can use light and flashes effects for video transitions. At the same moments, you need to additionally apply the camera shutter effect.
5. At the very end, select the background music and add it to the video in the CapCut application. You can stop at [the very track](https://www.youtube.com/watch?v=memDWTAvcXs) that is used in most videos, or you can find another one.
```python
def plot_square_loss_path(x_min, x_max, y_min, y_max, n, func, y, X, beta, coef=None):
    xx = np.linspace(x_min,x_max,n)
    yy = np.linspace(y_min,y_max,n)
    xxx, yyy = np.meshgrid(xx, yy)
    xx = xxx.reshape(-1)
    yy = yyy.reshape(-1)
    betas = np.hstack((xx[:,np.newaxis], yy[:, np.newaxis]))
    data = pd.DataFrame(np.hstack((xx[:, np.newaxis], yy[:,np.newaxis])), columns=['x', 'y'])
    S = np.array([func(y, X, beta) for beta in betas])
    data['z'] = S
    p = ggplot(data, aes('x', 'y')) \
        + geom_contourf(aes(z='z', fill='..level..'), bins=30) \
        + scale_fill_gradient('orange', 'red') \
        + geom_contour(aes(z='z', color='..level..'),bins=30) \
        + geom_point(x=beta[0], y=beta[1], size=5, color='blue') \
        + theme(legend_position='none') + coord_cartesian()

    if coef is not None:
        path = pd.DataFrame(coef, columns=['x', 'y'])
        p += geom_path(aes('x', 'y'), data=path, linetype='dashed', color='blue')

    return p
```

```python
def design_matrix(x, power=5):
    X = [x]
    for i in range(2, power + 1):
        x_powered = np.power(x, i)
        X.append(x_powered)
    return np.vstack(X).T
```

```python
"""
The Beale function has global minimum 0 at x=(3, 0.5)
"""
def beale(x1, x2):
    return np.log((1.5 - x1 + x1*x2)**2 + (2.25 - x1 + x1*x2**2)**2 + (2.625 - x1 + x1*x2**3)**2 + 2)
```

```python
def beale_3d(x1, x2):
    return x1, x2, beale(x1, x2)
```

```python
def grad_beale(x):
    x1, x2 = x
    a = (1.5 - x1 + x1*x2)
    b = (2.25 - x1 + x1*x2**2)
    c = (2.625 - x1 + x1*x2**3)
    offset = 2
    delta = a**2 + b**2 + c**2 + offset
    grad = np.array([(2*(x2 - 1)*a + 2*(x2**2 - 1)*b +2*(x2**3 - 1)*c)/delta, (2*x1*a + 4*x1*x2*b + 6*x1*x2**2*c)/delta])
    return grad
```

```python
def hess_beale(x):
    x1, x2 = x
    a = (1.5 - x1 + x1*x2)
    b = (2.25 - x1 + x1*x2**2)
    c = (2.625 - x1 + x1*x2**3)
    offset = 2

    a1 = (x2 - 1)
    a2 = x1
    b1 = (x2**2 - 1)
    b2 = 2*x1*x2
    c1 = (x2**3 - 1)
    c2 = 3*x1*x2**2

    a11 = 0
    a12 = 1
    a21 = 1
    a22 = 0
    b11 = 0
    b12 = 2*x2
    b21 = 2*x2
    b22 = 2*x1
    c11 = 0
    c12 = 3*x2**2
    c21 = 3*x2**2
    c22 = 6*x1*x2

    delta = a**2 + b**2 + c**2 + offset
    delta1 = 2*a*a1 + 2*b*b1 +2*c*c1
    delta2 = 2*a*a2 + 2*b*b2 + 2*c*c2

    delta11 = 2*(a1**2 + a*a11) + 2*(b1**2 + b*b11) + 2*(c1**2 + c*c11)
    delta12 = 2*(a2*a1 + a*a12) + 2*(b2*b1 + b*b12) + 2*(c2*c1 + c*c12)

    delta21 = 2*(a1*a2 + a*a21) + 2*(b1*b2 + b*b21) + 2*(c1*c2 + c*c21)
    delta22 = 2*(a2**2 + a*a22) + 2*(b2**2 + b*b22) + 2*(c2**2 + c*c22)

    H11 = (delta11*delta - delta1**2)/(delta**2)
    H12 = (delta12*delta - delta1*delta2)/(delta**2)
    H21 = (delta21*delta - delta2*delta1)/(delta**2)
    H22 = (delta22*delta - delta2**2)/(delta**2)
    return np.array([[H11, H12],[H21, H22]])
```

```python
def hessian(x, func, grad, eps=1e-8):
    n = x.size
    H = np.zeros((n,n))
    x1, x2 = x
    f_0 = func(x1, x2)
    for i in range(n):
        xx0 = 1.*x[i]
        x[i] = xx0 + eps
        df_0 = grad(x)
        H[i,:] = (df_0 - f_0)/eps
        x[i] = xx0
    return H
```

```python
def rotation_matrices(theta_x, theta_y, theta_z):
    Rx = [[1, 0, 0],
      [0, np.cos(theta_x), -np.sin(theta_x)],
      [0, np.sin(theta_x), np.cos(theta_x)]]
    Ry = [[np.cos(theta_y), 0, np.sin(theta_y)], 
      [0, 1, 0], 
      [-np.sin(theta_y), 0, np.cos(theta_y)]]
    Rz = [[np.cos(theta_z), -np.sin(theta_z), 0], 
      [np.sin(theta_z), np.cos(theta_z), 0], 
      [0, 0, 1]]
    return Rx, Ry, Rz
```

```python
def rotate(x, y, z, Rx, Ry, Rz):
    X = np.vstack((x, y, z)).T
    X_rotated = ((X.dot(Rx)).dot(Ry)).dot(Rz)
    return X_rotated
```

```python
def plot_2d_func(f, x1_min=-5, x1_max=5, x2_min=-5, x2_max=5, num=100):
    x1 = np.linspace(x1_min, x1_max, num=num)
    x2 = np.linspace(x2_min, x2_max, num=num)
    xx1, xx2 = np.meshgrid(x1, x2)
    xx1, xx2 = xx1.reshape(-1), xx2.reshape(-1)
    f_ = f(xx1, xx2)
    f_ = f_.reshape(-1)
    data = pd.DataFrame({'x1':xx1, 'x2':xx2, 'f':f_})
    p = ggplot(data) \
        + geom_contour(aes (x='x1', y='x2', z = 'f', color = '..level..'), bins=50, sampling='none') \
        + theme(legend_position='none') + scale_color_gradient(low='blue', high='red') + ggsize(700, 700)
    return p
```

```python
def plot_3d_func(f, x1_min=-5, x1_max=5, x2_min=-5, x2_max=5, num=100):
    x1 = np.linspace(x1_min, x1_max, num=num)
    x2 = np.linspace(x2_min, x2_max, num=num)
    xx1, xx2 = np.meshgrid(x1, x2)
    xx1, xx2 = xx1.reshape(-1), xx2.reshape(-1)
    x, y, z = f(xx1, xx2)
    X_rotated = rotate(x, y, z)
    data_3d = pd.DataFrame(X_rotated, columns=['x', 'y', 'z'])
    data_3d['longitudinal'] = np.tile(np.arange(num), num)
    data_3d['cross-sectional'] = np.repeat(np.arange(num), num)
    p = ggplot(data_3d) \
        + geom_path(aes('x', 'y', group='longitudinal'), alpha=0.5, color='gray') \
        + geom_path(aes('x', 'y', group='cross-sectional'), alpha=0.3, color='gray') \
        + theme(legend_position='none') + ggsize(700, 700) 
    return p
```

```python
def opt_path(opt, x, num_iter=350):
    path = np.zeros((num_iter, 2))
    path[0] = x
    for t in range(1, num_iter):
        x = opt.step(x, grad_beale)
        path[t] = x
    path = pd.DataFrame(path, columns=['x1', 'x2'])
    return path
```

```python
def opt_loss(x_opt, path):
    return np.sqrt(np.sum((path - x_opt)**2, axis=1))
```

```python
def opt_height(path, f=beale):
    x1, x2 = path[:, 0], path[:, 1]
    return f(x1, x2)
```

# 2. Linear regression
* Hypothesis function: $h_{\theta}(x)=\color{blue}{\theta_0}\cdot 1+\theta_1x_1+\ldots+\theta_nx_n$, where $\theta_0$ — is a bias term
* Our goal is to find $\theta=(\theta_0,\theta_1,\ldots,\theta_n)^T$ such that $h_{\theta}(x)$ is close to $y$
* Loss function — quadratic loss function: $$J(\theta)=\frac{1}{2}\sum_{i=1}^m\left(y^{(i)}-h_{\theta}(x^{(i)})\right)^2$$

## Algebraic approach

Compact algebraic form for hypothesis function of the linear regression $$\theta_0\cdot 1+\theta_1x_1^{(i)}+\ldots+\theta_nx_n^{(i)}=\theta^Tx^{(i)},$$ 
where $\theta=\left(\theta_0,\theta_1,\ldots,\theta_n\right)^T$ and $x^{(i)}=\left(1,x_1^{(i)},\ldots,x_n^{(i)}\right)^T$, for $i=1,\ldots,m$.

Linear regression model in the matrix form 
$$
\mathbf{y}=\mathrm{X}\theta\\
y = \left(
\begin{array}{c}
y_1 \\
y_2 \\
\vdots \\
y_m
\end{array}
\right)\in\mathbb{R}^{m},
\quad
\mathrm{X}=\left(
\begin{array}{cccccc}
1 & x^{(1)}_{1} & \cdots & x^{(1)}_{j} & \cdots & x^{(1)}_n \\
\vdots & \vdots & \ddots & x^{(i)}_j & \ddots & \vdots \\
1 & x^{(m)}_{1} & \cdots & x^{(m)}_{j} & \cdots & x^{(m)}_n
\end{array}
\right)\in\mathbb{R}^{m\times(n+1)},
\quad \theta=\left(
\begin{array}{c}
\theta_0 \\
\theta_1 \\
\vdots \\
\theta_n
\end{array}
\right)\in\mathbb{R}^{n+1}
$$
We denote $\mathbf{r}=\mathrm{X}\theta-\mathbf{y}$ - residual vector and get loss function in the matrix form  $\boxed{\;J(\theta)=\frac{1}{2}\mathbf{r}^T\mathbf{r}\;}$
Our goal is to find $\theta$ providing $\color{blue}{\text{minimum}}$ to the loss function.

Consider the next $\color{blue}{\text{algebraic trick}}$. How to solve the system $\mathbf{y}=\mathrm{X}\theta$ with respect to $\theta$?
If the design matrix $\mathrm{X}$ is square and non-singular, we can take the inverse matrix: $\mathrm{X}^{-1}: \mathrm{X}^{-1}\mathrm{X}=\mathrm{I}$, 
where $\mathrm{I}$ - identity matrix.
What can be done if $\mathrm{X}$ is not a square matrix? Let's multiply both sides of the equation by $\mathrm{X}^T$: $$\mathrm{X}^T\mathrm{X}\theta=\mathrm{X}^T\mathbf{y}$$
and then take the inverse matrix $$\boxed{\;\hat{\theta}=\left(\mathrm{X}^T\mathrm{X}\right)^{-1}\mathrm{X}^T\mathbf{y}\;}$$
Here we supposed, that $\mathrm{X}^T\mathrm{X}$ is not singular.
This estimate is called in mathematical statistics $\color{blue}{\text{least squares estimate}}$.

## Approach via derivatives
We can get a more strict derivation of the same formula. We want to find $\theta$ providing the minimum for the function
$$J(\theta)=\frac{1}{2}\left(\mathrm{X}\theta-y\right)^T\left(\mathrm{X}\theta-y\right)$$
Such $\theta$ is the root of the equation $\displaystyle \frac{\partial J(\theta)}{\partial \theta}=0$
Let's rewrite $J(\theta)$:
$$J(\theta)=\frac{1}{2}\left(\theta^T\mathrm{X}^T\mathrm{X}\theta-y^T\mathrm{X}\theta - \theta^T\mathrm{X}^Ty+y^Ty\right)$$
We are going to differentiate this expression with respect to $\theta$, so the last term is not relevant to our problem.

### The Gradient

Let $f:\mathbb{R}^{m\times n}\rightarrow\mathbb{R}$, then $\displaystyle \nabla_xf(x)=\left(
\begin{array}{ccc}
\frac{\partial f(x)}{\partial x_{11}} & \cdots & \frac{\partial f(x)}{\partial x_{1n}}\\
\vdots & \ddots & \vdots \\
\frac{\partial f(x)}{\partial x_{m1}} & \cdots & \frac{\partial f(x)}{\partial x_{mn}}
\end{array}
\right)\in\mathbb{R}^{m \times n}$
In particular if $f:\mathbb{R}^{m\times 1}\rightarrow\mathbb{R}$, then $\displaystyle \nabla_xf(x)=\left(
\begin{array}{c}
\frac{\partial f(x)}{\partial x_{1}} \\
\vdots \\
\frac{\partial f(x)}{\partial x_{m}} 
\end{array}
\right)\in\mathbb{R}^{m \times 1}$
$\color{blue}{\text{Note, that gradient has the same shape as } x}$.

### The Jacobian
Let $f:\mathbb{R}^{n}\rightarrow\mathbb{R}^m$, then $\displaystyle \frac{\partial f}{\partial x}=\left(
\begin{array}{ccc}
\frac{\partial f_1}{\partial x_1} & \cdots & \frac{\partial f_1}{\partial x_n}\\
\vdots & \ddots & \vdots \\
\frac{\partial f_m}{\partial x_1} & \cdots & \frac{\partial f_m}{\partial x_n}
\end{array}
\right)\in\mathbb{R}^{m \times n}$

Note that $\alpha=y^T\mathrm{X}\theta\in\mathbb{R}$ is a scalar value, so $\alpha=\alpha^T=\theta^T\mathrm{X}^Ty$ and
$$\frac{\partial J}{\partial \theta}=\frac{1}{2}\frac{\partial}{\partial\theta}\left(\theta^T\mathrm{X}^T\mathrm{X}\theta-2y^T\mathrm{X}\theta\right)$$
Denote $a^T=y^T\mathrm{X}\in\mathbb{R}^{1\times n}$ and $\mathrm{A}=\mathrm{X}^T\mathrm{X}\in\mathbb{R}^{n \times n}$, so we need to know the result of $\displaystyle \frac{\partial a^T\theta}{\partial\theta}$ and $\displaystyle \frac{\partial \theta^T\mathrm{A}\theta}{\partial \theta}$

Consider the first expression: let's take some index $k: 1\leq k \leq n$ and take the derivative with respect to $\theta_k$
$$\frac{\partial a^T\theta}{\partial \theta_k}=\frac{\partial}{\partial \theta_k}\sum_{i=1}^na_i\theta_i=a_k$$
taking this derivative with respect to each $\theta_k$, $k=1,\ldots,n$ we obtain the result 
$$\boxed{\; \frac{\partial a^T\theta}{\partial \theta}=a\;}$$

Consider now the gradient of the quadratic function:
$$\frac{\partial \theta^T A\theta}{\partial \theta_k}=\frac{\partial}{\partial \theta_k}\sum_{i=1}^n\sum_{j=1}^na_{ij}\theta_i\theta_j=\sum_{i=1}^na_{ik}\theta_i+\sum_{j=1}^na_{kj}\theta_j$$
and so
$$\boxed{\;\frac{\partial \theta^TA\theta}{\partial\theta}=A^T\theta+A\theta\;}$$
If $A$ is a symmetric matrix $A^T=A$, then
$$\boxed{\;\frac{\partial \theta^TA\theta}{\partial\theta}=2A\theta\;}$$

So we get for the quadratic loss function, that
$$\frac{\partial J}{\partial \theta}=\frac{1}{2}\frac{\partial}{\partial\theta}\left(\theta^T\mathrm{X}^T\mathrm{X}\theta-2y^T\mathrm{X}\theta\right)=\mathrm{X}^T\mathrm{X}\theta-\mathrm{X}^Ty=0$$
and if $\mathrm{X}^T\mathrm{X}$ is nonsingular we again get the least squares estimate: $$\boxed{\;\hat{\theta}=\left(\mathrm{X}^T\mathrm{X}\right)^{-1}\mathrm{X}^T\mathbf{y}\;}$$

### Numerical gradient
Let $\theta\in\mathbb{R}$ and $J(\theta)\in\mathbb{R}$. Consider two-sided function derivative 
$$\frac{d J(\theta)}{d\theta}=\lim_{\varepsilon\rightarrow0}\frac{J(\theta+\varepsilon)-J(\theta-\varepsilon)}{2\varepsilon}$$
We can generalize the above formula for the case of $\theta\in\mathbb{R}^{n}$:
$$\frac{\partial}{\partial\theta_1}J(\theta)=\lim_{\varepsilon\rightarrow0}\frac{J(\theta_1+\varepsilon,\theta_2,\ldots,\theta_n)-J(\theta_1-\varepsilon,\theta_2,\ldots,\theta_n)}{2\varepsilon};\\
\frac{\partial}{\partial\theta_2}J(\theta)=\lim_{\varepsilon\rightarrow0}\frac{J(\theta_1,\theta_2+\varepsilon,\ldots,\theta_n)-J(\theta_1,\theta_2-\varepsilon,\ldots,\theta_n)}{2\varepsilon};\\
\vdots\\
\frac{\partial}{\partial\theta_n}J(\theta)=\lim_{\varepsilon\rightarrow0}\frac{J(\theta_1,\theta_2,\ldots,\theta_n+\varepsilon)-J(\theta_1,\theta_2,\ldots,\theta_n-\varepsilon)}{2\varepsilon}$$

## SGD, classic

See the properties of GD convergence [here](https://www.slideshare.net/SebastianRuder/optimization-for-deep-learning?ref=ruder.io)

Stochastic Gradient Descent update rule:
$$\boxed{\;\theta_{t+1} = \theta_{t} -\eta g_t \;}$$

```python
from typing import Callable, Optional, Tuple

Pair = Tuple[float, float]

class SGD:
    def __init__(self, eta: float = 0.1) -> None:
        self.eta: float = eta

    def step(self, x: Pair, grad: Callable[[Pair], float], eta: Optional[float] = None) -> Pair:
        if eta is None:
            eta = self.eta
        return x - eta * grad(x)
```

## Adam, 2015

[Link](https://arxiv.org/pdf/1412.6980.pdf) to the paper

• Get the gradients $\displaystyle g_{t,i}=\nabla_{\theta}J(\theta_{t,i})$

• Calculate the averages:

  • $\displaystyle v_t=\beta_1 v_{t-1}+(1-\beta_1)g_t$

  • $\displaystyle G_t=\beta_2 G_{t-1}+(1-\beta_2)g_t^2$

• Perform bias correction:

  • $\displaystyle \hat{v}_t=\frac{v_t}{1-\beta_1^t}$

  • $\displaystyle \hat{G}_t=\frac{G_t}{1-\beta_2^t}$

• Update estimates $\displaystyle \theta_{t+1}=\theta_t-\frac{\eta}{\sqrt{\hat{G}_t+\varepsilon}}\hat{v}_t$  

```python
class Adam:
    def __init__(self, eta=0.1, beta1=0.9, beta2=0.999, eps=1e-8):
        self.eta = eta
        self.beta1 = beta1
        self.beta2 = beta2
        self.eps = eps
        self.v = np.zeros(2)
        self.G = np.zeros(2)
        self.t = 0

    def step(self, x, grad, eta=None, beta1=None, beta2=None):
        if eta is None:
            eta = self.eta
        if beta1 is None:
            beta1 = self.beta1
        if beta2 is None:
            beta2 = self.beta2

        self.t += 1
        g = grad(x)
        self.v = beta1*self.v + (1 - beta1)*g
        self.G = beta2*self.G + (1 - beta2)*g**2
        v = self.v/(1 - beta1**self.t)
        G = self.G/(1 - beta2**self.t)
        x = x - eta/(np.sqrt(G) + self.eps)*v
        return x
```

## Newton's method, classic

Consider the Taylor series expantion

$$J(\theta)\eqsim J(\theta_0)+(\theta-\theta_0)^T\nabla_{\theta}J(\theta_0)+\frac{1}{2}(\theta-\theta_0)^TH(\theta-\theta_0)$$

• Get the gradients $\displaystyle g=\nabla_{\theta}J(\theta)$

• We need to find the critical points of $J$, i.e. roots of $\displaystyle g(\theta)=\nabla_{\theta}J(\theta) = 0$

• The Newton-Raphson update formula for $g(\theta)$ is: 
$\theta \leftarrow \theta - (Jacobian(g(\theta)))^{-1} * g(\theta)$

• Compute the Hessian $\displaystyle H=\nabla^2_{\theta}J(\theta)$ 

• Compute the inverse matrix $\left(H+\alpha I\right)^{-1}$

• Update the estimates $\displaystyle \theta \leftarrow \theta -\left(H+\alpha I\right)^{-1}g$

```python
"""
https://arxiv.org/pdf/1707.08028.pdf
"""
def pt_inverse(A, mu):
    w, V = np.linalg.eigh(A)
    w = np.array([np.abs(wi) if np.abs(wi)> mu else mu for wi in w])
    w = np.diag(w)
    return V.dot(np.linalg.inv(w).dot(V.T))
```

```python
class Newton:
    def __init__(self, func=beale, eta=.1, alpha=1., beta=0.9, eps=1e-8):
        self.alpha = alpha
        self.beta = beta
        self.eta = eta
        self.eps = eps
        self.func = func

    def step(self, x, grad, hess=hess_beale, alpha=None):
        if alpha is None:
            alpha = self.alpha
        H = hess(x)
        g = grad(x)
        H_ = pt_inverse(H, 1e-8)
        eta = self.eta
        x_new = x - eta*H_.dot(g)
        x1, x2 = x
        f_x = self.func(x1, x2)
        x1, x2 = x_new
        f_new = self.func(x1, x2)
        while f_new > f_x - alpha*eta*g[:, np.newaxis].T.dot(H_).dot(g):
            eta = eta*self.beta
            x_new = x - eta*H_.dot(g)*2
            x1, x2 = x_new
            f_new = self.func(x1, x2)
        return x_new
```

## CyclicalSGD

We need to keep track of the current iteration and adjust the learning rate according to a predefined schedule from eta_min to eta_max.

```python
from typing import Callable, Optional, Tuple
from math import cos, pi

Pair = Tuple[float, float]

class CyclicalSGD:
    def __init__(self, eta_min: float = 0.01, eta_max: float = 1, cycle_length: int = 50) -> None:
        self.eta_min: float = eta_min
        self.eta_max: float = eta_max
        self.cycle_length: int = cycle_length
        self.iteration: int = 0

    def step(self, x: Pair, grad: Callable[[Pair], Pair], eta: Optional[float] = None) -> Pair:
        if eta is None:
            eta = self.get_cyclical_learning_rate()
        self.iteration += 1
        return x - eta * grad(x)

    def get_cyclical_learning_rate(self) -> float:
        cycle_position = self.iteration % self.cycle_length
        return self.eta_min + 0.5 * (self.eta_max - self.eta_min) * (1 + cos(pi * cycle_position / self.cycle_length))
```

### SWA weights updates

Perform exponential averaging of model weights to get expected values

$$\boxed{\displaystyle \theta_{\text{SWA}} \leftarrow \frac{\theta_{\text{SWA}} \cdot T+\theta}{T+1}}$$

### SWAG-Diagonal

Estimate diagonal elements of the covariance matrix

$$\displaystyle \overline{\theta^2}=\frac{1}{T}\sum_{i=1}^T\theta_i^2$$
$$\displaystyle \Sigma_{\text{diag}}=\textrm{diag}\left(\overline{\theta^2} - \theta_{\text{SWA}}^2\right)$$

We can consider models' weights to be represented by a multidemensional Normal distribution
$$\displaystyle \mathcal{N}\left(\theta_{\text{SWA}}, \Sigma_{\text{diag}}\right)$$

### SWAG: Low Rank plus Diagonal Covariance

Consider the full covariance matrix

$$\displaystyle \Sigma = \frac{1}{T-1}\sum_{i=1}^T\left(\theta_i - \theta_{\text{SWA}}\right)\left(\theta_i - \theta_{\text{SWA}}\right)^T$$

$$\displaystyle \Sigma = \frac{1}{T-1}\sum_{i=1}^T\left(\theta_i - \overline{\theta_i}\right)\left(\theta_i - \overline{\theta_i}\right)^T=\frac{1}{T-1}DD^T$$

Consider only $K$ last parameter updates $i = T-K+1 ,..., T$
$$\displaystyle D_i=\left(\theta_i - \overline{\theta_i}\right)$$

and take the low-rank approximation of the covariance matrix 
$$\Sigma_{\text{low-rank}}=\frac{1}{K-1}\widehat{D}\widehat{D}^T$$

Combining together diagonal matrix and low-rank approximation, we obtain the next distribution for models' parameters
$$\displaystyle \mathcal{N}\left(\theta_{\text{SWA}}, \frac{1}{2}\left(\Sigma_{\text{diag}} + \Sigma_{\text{low-rank}}\right)\right)$$

Model parameters can be sampled using this formula
$$\boxed{\widetilde{\theta}=\theta_{\text{SWA}} + \frac{1}{\sqrt{2}}\cdot \Sigma_{\text{diag}}^{\frac{1}{2}}z_1+\frac{1}{\sqrt{2(K-1)}}\widehat{D}z_2, \text{  where } z_1\sim \mathcal{N}\left(0, I_d\right), z_2\sim \mathcal{N}\left(0, I_K\right)}$$

# 4. Linear Regression from the Probabilistic point of view

From the probabilistic point of view the linear regression model can be interpreted in the following way: $$y^{(i)}=\theta^Tx^{(i)}+\varepsilon^{(i)}, \varepsilon^{(i)}\sim N(0,\sigma^2)$$
Univariate normal distribution $$p(x;\mu,\sigma)=\frac{1}{\sqrt{2\pi}\sigma}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)$$

```python
mu = 0
sigma = 1
n = 1000
x = norm.rvs(loc=mu, scale=sigma, size=n) # Random variates
data = pd.DataFrame({'x':x, 'f':norm.pdf(x)}) # Probability Density Function
```

Then $\displaystyle (y^{(i)}\;|\;x^{(i)};\theta)\sim N(\theta^Tx^{(i)},\sigma^2)$. Suppose, the observations are independent, then we can define the $\color{blue}{\text{likelihood function}}$:  
$$L(\theta)=L(\theta;\mathrm{X},\mathbf{y})=p(\mathbf{y}\,|\,\mathrm{X},\theta)=\prod_{i=1}^mp(y^{(i)}\,|\,x^{(i)};\theta)$$
More handy function to work with is the $\color{blue}{\text{log-likelihood function}}$ $$\ell(\theta)=\color{blue}{\log} L(\theta)=m\log\frac{1}{\sqrt{2\pi}\sigma}-\frac{1}{\color{blue}2\sigma^2}\color{blue}{\sum_{i=1}^m\left(y^{(i)}-\theta^Tx^{(i)}\right)^2}$$
So, we get, that $\color{blue}{\text{maximizing}}$ the likelihood function is equivalent to $\color{blue}{\text{minimizing}}$ the quadratic loss function $J(\theta)$

## ML and MAP estimates
* From the $\color{blue}{\text{Chain Law of probability}}$ obtain
$$p\Big(w,y,x\Big)=p\Big(w\,\Big|\,y,x\Big)p\Big(y \,\Big|\, x\Big)p\Big(x\Big)$$
or
$$p\Big(w,y,x\Big)=p\Big(y\,\Big|\,w,x\Big)p\Big(w \,\Big|\, x\Big)p\Big(x\Big)$$
* and finally we get $\color{blue}{\text{Bayes theorem}}$ statement $\color{magenta}{\text{(with simplification due to indenedence of x and w in our case)}}$:
$$p\Big(w\,\Big|\,y,x\Big)=p^{-1}\Big(y\,\Big|\, x\Big)p\Big(y\,\Big|\,w,x\Big)\color{magenta}{p\Big(w \,\Big|\, x\Big)}=p^{-1}\Big(y\,\Big|\, x\Big)p\Big(y\,\Big|\,w,x\Big)\color{magenta}{p\Big(w \Big)},$$
where:
 * $\ell\Big(w\,\Big|\,y,x\Big)=p\Big(y\,\Big|\,w,x\Big)$ - $\color{blue}{\text{likelihood function}}$;
 * $p\Big(w\Big)$ - $\color{blue}{\text{prior distribution}}$ for parameters $w$;
 * $p\Big(w\,\Big|\,y,x\Big)$ - $\color{blue}{\text{posterior distribution}}$;
 * $p\Big(y \,\Big|\, x\Big)$ - $\color{blue}{\text{model evidence}}$ - information, contained in examples.
  
**Maximum Likelihood** estimate: 
$$\boxed{\;w_{ML}=\arg\max_{w}\Bigg[\ell\Big(w\,\Big|\,y,x\Big)\Bigg]=\arg\max_w\log\Bigg[\ell\Big(w\,\Big|\,y,x\Big)\Bigg]\;}$$
**Maximum A Posteriori** estimate: 
$$\boxed{\;w_{MAP}=\arg\max_w\Bigg[p\Big(w\,\Big|\,y,x\Big)\Bigg]=\arg\max_w\log \Bigg[p\Big(w\,\Big|\,y,x\Big)\Bigg]\;}$$

Recall the algebraic approach: $$\boxed{\;\hat{\theta}=\left(\mathrm{X}^T\mathrm{X} + \alpha \mathrm{I}\right)^{-1}\mathrm{X}^T\mathbf{y}\;}$$

```python
def MAP(X, y, alpha=0.0):
    m = X.shape[1]
    I = np.identity(m)
    Rxx = np.dot(X.T, X)
    Ryx = np.dot(X.T, y)
    if alpha == 0.0:
        return np.dot(np.linalg.inv(Rxx), Ryx)
    elif alpha > 0.0:
        return np.dot(np.linalg.inv(Rxx + alpha * I), Ryx)
```

# L2 regularization

# 5. $L_2$ regularization:

* **Posterior density** in Gaussian case:
$$p(\theta\,|\,\mathbf{y},\mathrm{X}) \propto \exp\thinspace\left(\underbrace{-\frac{1}{2\sigma^2}\sum_{i=1}^m(y^{(i)}-\theta^Tx^{(i)})^2}_{\text{log-likelihood}}-\underbrace{\frac{1}{2\sigma_{\theta}^2}\|\theta\|^2}_{\text{log-prior}}\right)$$

* Gaussian case, **regularized cost function** for MAP: 
$$J(\theta)=\underbrace{\frac{1}{2}\sum_{i=1}^m(y^{(i)}-\theta^Tx^{(i)})^2}_{\text{Empirical risk}}+\underbrace{\frac{\alpha}{2}\|\theta\|^2}_{\text{Penalty}},$$
where $\alpha=\frac{\sigma^2}{\sigma_{\theta}^2}$

* Closed form solution: 
    * Solve the equation: $\displaystyle{\frac{\partial J}{\partial \theta}=-2\mathrm{X}^T(y-\mathrm{X}\theta)+\alpha\theta=0}$
    * Estimate: $\displaystyle{\hat{\theta}=(\mathrm{X}^T\mathrm{X}+\alpha\mathrm{I})^{-1}\mathrm{X}^Ty}$

## Linear Regression Model for regression task

* Estimate coefficients $\hat{\theta}$;
* Make a model prediction $\hat{y}^{(i)}=\hat{\theta}^Tx^{(i)}$;
* Score $\displaystyle{\mathrm{R}^2=1-\frac{\sum_{i=1}^m(y^{(i)}-\hat{y}^{(i)})^2}{(y^{(i)}-\bar{y})^2}}$, where $\displaystyle{\bar{y}=\frac{1}{m}\sum_{i=1}^my^{(i)}}$

```python
class RidgeLinearRegression(object):
    def __init__(self, alpha=0.0):
        self.alpha_ = alpha

    def fit(self, X, y, alpha=None):
        if alpha == None:
            alpha = self.alpha_
        self.coef_ = MAP(X, y, alpha)
        return self.predict(X)

    def predict(self, X):
        self.y_ = np.dot(X, self.coef_)
        return self.y_

    def score(self, X, y):
        return 1. - np.sum((y-self.predict(X))**2)/np.var(y)/len(y)
```

## Linear Regression for Polynomial model

* Polynomial model $\displaystyle{y=\mathrm{P}_n(\theta, x)}$, where $\displaystyle{\mathrm{P}_n(\theta, x)=\theta_0+\sum_{i=1}^n\theta_ix^i}$
* Input vector in matrix form $x=(1,x,\ldots,x^n)$
* The same linear regression model $y=\theta^Tx+\varepsilon$

```python
def poly(c, x):
    y = np.zeros_like(x)
    for i in np.arange(len(c)):
        y = y + c[i] * x**i
    return y
```

# L1 regularization

# 6. Connection to $L_1$ regularization:

**Posterior density** in Gaussian case with Laplace prior:
$$p(\theta\,|\,\mathbf{y},\mathrm{X}) \propto \exp\thinspace\left(\underbrace{-\frac{1}{2\sigma^2}\sum_{i=1}^m(y^{(i)}-\theta^Tx^{(i)})^2}_{\text{log-likelihood}}-\underbrace{\frac{1}{\sigma_{\theta}}|\theta|}_{\text{log-prior}}\right)$$

Gaussian case, **regularized cost function** for MAP: 
$$J(\theta)=\underbrace{\frac{1}{2}\sum_{i=1}^m(y^{(i)}-\theta^Tx^{(i)})^2}_{\text{Empirical risk}}+\underbrace{\alpha|\theta|}_{\text{Penalty}},$$
where $\alpha=\frac{\sigma^2}{\sigma_{\theta}}$

No closed-form solution in this case, but can be interpreted as ($\color{green}{\text{Euler-Lagrange equation}}$):

*minimize $J_0(\theta)$ subject to the constraint $\displaystyle{\sum_{i=0}^n|\theta_j|\leq \eta}$ for some $\eta$, where $J_0(\theta)$ stands for cost function of original (unconstrained) maximum likelihood.*
    
Suppose $\theta_j>0$, then 
$$\frac{\partial J}{\partial \theta_j}=\sum_{i=1}^m\left(y^{(i)}-\sum_{k=1}^n\theta_kx_{k}^{(i)} \right)x_{j}^{(i)}-\alpha=\\=\sum_{i=1}^m\left(y^{(i)}-\sum_{\color{blue}{k\neq j}}\theta_kx_{k}^{(i)} - \color{blue}{\theta_jx_{j}^{(i)}}\right)x_{j}^{(i)}-\alpha=\\=\sum_{i=1}^m\left(y^{(i)}-\sum_{k\neq j}\theta_kx_{k}^{(i)}\right)x_{j}^{(i)}-\color{blue}{\theta_j\sum_{i=1}^m\left(x_{j}^{(i)}\right)^2}-\alpha=0$$

and 

$$\displaystyle{\color{blue}{\hat{\theta}_j}=\underbrace{\frac{\sum_{i=1}^m\left(y^{(i)}-\sum_{k\neq j}\theta_kx_{k}^{(i)}\right)x_{j}^{(i)}}{\color{blue}{\sum_{i=1}^m\left(x_{j}^{(i)}\right)^2}}}_{\hat{\theta}_j^{ML}}-\frac{\alpha}{\color{blue}{\sum_{i=1}^m\left(x_{j}^{(i)}\right)^2}}}$$

which (keeping in mind, that $\theta_j>0$) means that we admit a solution only in case if $\displaystyle{\hat{\theta}_j^{ML}>\frac{\alpha}{\sum_i\left(x_{j}^{(i)}\right)^2}}$

Combining both variants we obtain
    $$\displaystyle{\hat{\theta}_j=\left\{\begin{array}{ll}
    \hat{\theta}_j^{ML}-\eta_j\,\mathrm{sign}(\hat{\theta}_j^{ML}), & |\hat{\theta}_j^{ML}|>\eta_j \\
    0, & |\hat{\theta}_j^{ML}|\leq\eta_j
    \end{array}\right.}$$
    where $\displaystyle{\eta_j=\frac{\alpha}{\sum_i\left(x_{j}^{(i)}\right)^2}}$

```python
import numpy as np
import matplotlib.pyplot as plt

# Create meshgrid for contour plots
x = np.linspace(-1.5, 1.5, 100)
y = np.linspace(-1.5, 1.5, 100)
X, Y = np.meshgrid(x, y)

# Define contour function (objective function)
Z = (X-0.8)**2 + (Y-1.2)**2

fig, ax = plt.subplots(1, 2, figsize=(10, 5))

# Ridge Regression (L2 regularization)
ax[0].contour(X, Y, Z, levels=15, cmap='viridis')
ax[0].add_artist(plt.Circle((0, 0), 0.55, color='r', fill=False, linewidth=3.))
ax[0].add_artist(plt.Circle((0.32, 0.46), 0.05, color='black', fill=True))
ax[0].set_xlim(-1.5, 1.5)
ax[0].set_ylim(-1.5, 1.5)
ax[0].set_aspect('equal')
ax[0].set_title('Ridge Regression (L2)')

# Lasso Regression (L1 regularization)
ax[1].contour(X, Y, Z, levels=25, cmap='viridis')
r = 0.58
ax[1].add_artist(plt.Polygon([[-r, 0], [0, r], [r, 0], [0, -r]], color='r', fill=False, linewidth=3.))
ax[1].add_artist(plt.Circle((0., 0.58), 0.05, color='black', fill=True))
ax[1].set_xlim(-1.5, 1.5)
ax[1].set_ylim(-1.5, 1.5)
ax[1].set_aspect('equal')
ax[1].set_title('Lasso Regression (L1)')

plt.tight_layout()
plt.show()
```

## How the soft threshold works

```python
def soft_treshold(x, a):
    if abs(x) > a:
        return x - a*np.sign(x)
    else:
        return 0
```

```python
z = norm.rvs(size=1000)
zt = np.array([soft_treshold(zi, 1.0) for zi in z])
data = {'z': z, 'zt': zt}
```

```python
def coord_descent(X, y, alpha=0.1, max_iter=100):
    w = np.zeros(X.shape[1])
    for _ in range(max_iter):
        for j in range(len(w)):
            wj = deepcopy(w)
            wj[j] = 0.
            rj = y - np.dot(X, wj)
            x = np.dot(X[:,j], rj)
            a = alpha*X.shape[0]
            w[j] = soft_treshold(x, a)/(X[:,j]**2).sum()
    return w
```

```python
def MAP_L1(X, y, alpha=0.0):
    Rxx = np.dot(X.T, X)
    Ryx = np.dot(X.T, y)
    if alpha == 0.0:
        return np.dot(np.linalg.inv(Rxx), Ryx)
    elif alpha > 0.0:
        return coord_descent(X, y, alpha)
```

```python
class LassoLinearRegression(object):
    def __init__(self, alpha=0.1):
        self.alpha_ = alpha

    def fit(self, X, y, alpha=None):
        if alpha == None:
            alpha = self.alpha_
        self.coef_ = MAP_L1(X, y, alpha)
        return self.predict(X)

    def predict(self, X):
        self.y_ = np.dot(X, self.coef_)
        return self.y_

    def score(self, X, y):
        return 1. - np.sum((y-self.predict(X))**2)/np.var(y)/len(y)
```

# Deep Double Descent

# 7. [Deep Double Descent](https://openai.com/blog/deep-double-descent/)
Double descent phenomenon occurs in CNNs, ResNets, and transformers: performance first improves, then gets worse, and then improves again with increasing model size, data size, or training time. This effect is often avoided through careful regularization. While this behavior appears to be fairly universal, we don’t yet fully understand why it happens, and view further study of this phenomenon as an important research direction.
